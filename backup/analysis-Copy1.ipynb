## IMPORTS

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import pandas as pd
import numpy as np
```

## Import data

```python
data_path = Path('../data/cleaned')
data_files = list(data_path.glob('*.csv'))
print(f'{len(data_files)} files found:\n')
for file in data_files:
    print(f'\t-{file.name}')
```

```python
for file in data_files:
    if 'commits' in file.name:
        commits_df = pd.read_csv(data_path/file.name, parse_dates=['author_date', 'commit_date'])
    elif 'issues' in file.name:
        issues_df = pd.read_csv(data_path/file.name, parse_dates=['created_at', 'closed_at', 'updated_at'])
    elif 'info' in file.name:
        info_df = pd.read_csv(data_path/file.name, parse_dates=['created_at', 'updated_at', 'pushed_at' ])
```

```python
# Check repo number:

expected_repos = 5
actual_repos = commits_df['repo_name'].nunique()
assert actual_repos == expected_repos, f"Expected {expected_repos} repositories, but found {actual_repos}"
print(f"✓ Repository count test passed: {actual_repos} repos")
```

## Repo summary

```python
# Group issue data by repo_name and perform summary stats:

issues_summary = issues_df.groupby('repo_name').agg({
    'issue_id': 'count',
    'issue_age_days': ['mean', 'median']
})

issues_summary = issues_summary.reset_index() 
issues_summary.columns = (
    f'{top_level}_{bottom_level}' for top_level, bottom_level in issues_summary.columns
)
issues_summary
```

```python
# Group commit data by repo_name and perform aggregation:

commits_summary = commits_df.groupby('repo_name').agg({
    'sha': 'count'
})
commits_summary = commits_summary.reset_index().rename(columns ={'sha': 'commit_sha_count','repo_name': 'repo_name_'})
commits_summary
```

```python
# Group info data by repo_name and perform aggregation:

info_summary = info_df.groupby('repo_name').agg({
    'stars': 'sum'
})

info_summary = info_summary.reset_index().rename(columns={'stars': 'starsgazers_sum'})
info_summary
```

```python
# join all the summaries into one summary:

summary_df = pd.merge(issues_summary, commits_summary, on='repo_name_', how='inner').rename(columns={'repo_name_': 'repo_name',
'issue_age_days_mean': 'avg_issue_age/days',
 'issue_age_days_median': 'median_issue_age/days'                                                  })
summary_df = pd.merge(summary_df, info_summary, on='repo_name', how='inner' )
summary_df
```

```python
#The number of unique contributors:
commits_df['author_name'].nunique()
```

```python
#Top 5 contributors:
print('--- Top five Contributors ---')
commits_df['author_name'].value_counts().head(5)
```

```python
# Check the proportion of closed versus open issues:
issues_df['state'].value_counts(normalize=True)
```

```python
#create new column, closed_issues to hold the time it took to close issues (only actually closed issues):
closed_issues = issues_df[['repo_name', 'time_to_close_hours']].dropna(subset=['time_to_close_hours'])

#Group time to close issues by 'repo_name' and calculate 'average_time_to_close issues' for each rep:
closed_issues = closed_issues.groupby('repo_name').agg({
    'time_to_close_hours': 'mean',
}).reset_index().rename(columns={'time_to_close_hours': 'average_time_to_close_hours'})

closed_issues # Its already sorted in ascending order
```

```python
#Create resolve_time_cmpr to hold the time it took to close issues and the author association of closed issues:
resolve_time_cmpr = issues_df[['repo_name', 'time_to_close_hours', 'author_association']].dropna(subset=['time_to_close_hours'])

#Group resolve_time_cmpr by repo_name and author_association and then find the mean and median time for each author_association and repo:
resolve_time_cmpr = resolve_time_cmpr.groupby(['repo_name','author_association']).agg({
    'time_to_close_hours':['mean', 'median']
})

# Flatten new columns:
resolve_time_cmpr.columns = (
    f'{top_level}_{bottom_level}' for top_level, bottom_level in resolve_time_cmpr.columns
)

resolve_time_cmpr = resolve_time_cmpr.rename(columns={
    'time_to_close_hours_mean': 'avg_time_to_close',
    'time_to_close_hours_median': 'median_time_to_close'
})
```

```python
#create commits_by_date to store repo_name, author_date and message_length:
commits_by_date = commits_df[['repo_name', 'author_date','message_length' ]].copy()

commits_by_date['day_of_the_week'] = commits_by_date['author_date'].dt.day_name()

commits_by_date['author_date'] = commits_by_date['author_date'].dt.date

#Create commits_per_day_df to hold the count of the number of commits per day by repo_name:
commits_per_day = commits_by_date.groupby(['repo_name', 'author_date']).size()

commits_per_day_df = commits_per_day.reset_index(name='number_of_commits')

commits_per_day_df = commits_per_day_df.set_index(['repo_name', 'author_date'])

commits_per_day_df.sort_values(by='number_of_commits', ascending=False)
```

```python
#Which days had the most commits:
day_analysis = commits_by_date.groupby(['repo_name', 'day_of_the_week']).agg({
    'day_of_the_week': 'count'
})

day_analysis = day_analysis.rename(columns={
    'day_of_the_week': 'number_of_commits_on_day'
})
```

```python
#Message Length analysis:
commit_msg_analysis = commits_by_date.copy()
commit_msg_analysis = commit_msg_analysis.groupby(['repo_name']).agg({
    'message_length': ['mean', 'median']
})

# Flatten new columns:
commit_msg_analysis.columns = (
    f'{top_level}_{bottom_level}' for top_level, bottom_level in commit_msg_analysis.columns
)

commit_msg_analysis = commit_msg_analysis.rename(columns={
    'message_length_mean': 'avg_message_length',
    'message_length_median': 'median_message_length'
})

    
commit_msg_analysis.sort_values(by= 'avg_message_length', ascending=False)
```

## VISUALIZATIONS

```python
#Prepare the data in issues_df['closed_at', 'created_at'] to be graphed:
#Store repo_name, created_at, closed_at, and state in monthly_graoh:
monthly_graph = issues_df[['repo_name', 'created_at', 'closed_at', 'state']].copy()

#Convert closed_at and created_at to year-month format:
monthly_graph['closed_at'] = monthly_graph['closed_at'].dt.to_period('M')

monthly_graph['created_at'] = monthly_graph['created_at'].dt.to_period('M')
```

```python
#Isolate and count open issues per month:
open_issues = monthly_graph[monthly_graph['state'] == 'open']

open_issues_by_month = open_issues.groupby('created_at').size().reset_index(name='open_issues_count')

open_issues_by_month
```

```python
#Isolate and count closed issues per month:
closed_issues = monthly_graph[monthly_graph['state'] == 'closed']

closed_issues = closed_issues.dropna()

closed_issues_by_month = closed_issues.groupby('closed_at').size().reset_index(name='closed_issues_count')

closed_issues_by_month
```

### Graphing

#### Issue Activity over time

```python
#Prepare the graphing area:
sns.set_style('dark')
plt.figure(figsize=(10,8.4))
axis = plt.gca()

#plot open issues:
open_issues_by_month['month_str'] = open_issues_by_month['created_at'].astype(str)
sns.lineplot(
    ax=axis,
    data=open_issues_by_month,
    x='month_str',
    y='open_issues_count',
    label='Issues opened',
    marker='o'
)

#plot closed issues:
closed_issues_by_month['month_str'] = closed_issues_by_month['closed_at'].astype(str)
sns.lineplot(
    ax=axis,
    data=closed_issues_by_month,
    x='month_str',
    y='closed_issues_count',
    label='Issues Closed',
    marker='o'
    
)
plt.title('Issue Activity Over Time')
plt.xlabel('Month')
plt.ylabel('Number of Open/Closed Issues')

plt.show
```

#### Project Health Comparison

```python
#Prepare Graphing area:
sns.set_style('darkgrid')
plt.figure(figsize=(10,6))
ax = plt.gca()

#Plot chart:
sns.barplot(
    data=summary_df,
    y='avg_issue_age/days',
    x='repo_name',
    ax=ax
)
plt.title('Average Issue Age by Project Repository')
plt.xlabel('Project')
plt.ylabel('Average Issue Age(days)')
```

#### Contributer Distribution

```python
#Prepare data to be plotted:
top_contributor_dist = (
    commits_df.groupby(['repo_name', 'author_name'])
        .size()
        .reset_index(name='commit_count')
        .sort_values(['repo_name', 'commit_count'], ascending=[True, False])
)

top_ten_contributors = top_contributor_dist.groupby('repo_name').head(10)
```

```python
plt.figure(figsize=(30,26))
sns.set_style('darkgrid')

sns.barplot(
    data = top_ten_contributors,
    x = 'author_name',
    y = 'commit_count',
    hue= 'repo_name'
)

plt.xlabel('Author Name')
plt.ylabel('Number of Commits')
plt.title('Top 10 Contributors per Repo')
plt.legend(title='Repository')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

#### Issue Age Distribution

```python
#Check skewness of data:
sns.set_style('dark')
plt.figure(figsize=(10,6))
sns.histplot(
    data=issues_df,
    x='issue_age_days',
    kde=True,
)
```

```python
# Log Transform data using ln(x+1)
issues_df['issue_age_days_log'] = np.log1p(issues_df['issue_age_days'])

#Prepare plotting Area:
sns.set_style('dark')
plt.figure(figsize=(10,6))

#Plot Data
sns.histplot(
    data=issues_df,
    x='issue_age_days_log',
    kde=True,
)
plt.title('Issue Age Distribution')
plt.xlabel('Log of Issue Age')
plt.ylabel('Count')
```

#### Commits by Day of Week

```python
#Prepare Data:
day_plot = day_analysis.reset_index()

#Prepare plotting area:
sns.set_style('whitegrid')
plt.figure(figsize=(12,7))

#plot data:
sns.barplot(
    data=day_analysis,
    x='day_of_the_week',
    y='number_of_commits_on_day',
    hue='repo_name'
)

plt.title('Distribution of Commits Across Repos by day of the week')
plt.xlabel('Day')
plt.ylabel('Number of Commits')
```
